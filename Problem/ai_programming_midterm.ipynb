```python

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50  # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
    "HR Manager",
    "Software Developer",
    "Accountant",
    "Marketing Executive",
    "Sales Executive",
    "Backend Developer",
    "HR Executive",
    "Financial Analyst",
    "Digital Marketer",
    "Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame({
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_salary": emp_income,
    "role": emp_roles
})

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

```python
salary = np.array([emp_income])
max_salary = np.max(salary)
min_salary = np.min(salary)
mean_salary = np.mean(salary)
print('maximum salary: ', max_salary)
print('minimum salary: ', min_salary)
print("average salary: ",mean_salary)

ages = np.array([emp_ages])
mean_age = np.mean(ages)
print("average age: ", mean_age)
```

```python
emp_age = np.array(emp_ages)
emp_salary = np.array(emp_income)
aged_b_25_t_40 = np.where((emp_age >= 25) & (emp_age <= 40))
more_inc_than_50k = np.where(emp_salary >= 50000)
count = np.count_nonzero(more_inc_than_50k)
total_emp_age = len(aged_b_25_t_40[0])
total_emp_salary =len(more_inc_than_50k[0])

print('are aged between 25 and 40: ',total_emp_age)
print('have salary greater than 50,000: ',total_emp_salary)
print("total employs: ",total_emp_age + total_emp_salary)
```

```python
emp_age = np.array(emp_ages)
mean_ages = np.mean(emp_age)
adjust_ages = np.where(emp_age > mean_age , emp_age + 2, emp_age + 4)


print('original array: ',emp_age)
print('updated array: ',adjust_ages)
print("average age: ",mean_age)
```

```python
emp_role = np.array(emp_roles)
emp_salary = np.array(emp_income)
tech_roles = np.where((emp_role == "Backend Developer") | (emp_role == "Digital Marketer") | (emp_role == "Software Developer" ))
present_salary = emp_salary[tech_roles]
salary_increase = np.where(present_salary > 0, (present_salary * 1.08),0)
total_present_salary = np.sum(present_salary)
new_total_salary = np.sum(salary_increase)


print("new total salary expense: ",new_total_salary)
print('overall increase in expense: ', new_total_salary - total_present_salary)
```
