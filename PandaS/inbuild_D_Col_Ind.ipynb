```python
import pandas as pd
```

```python
my_list = [['Alice', 25],['Bob', 30], ['Charlie', 28]]

list_df = pd.DataFrame(my_list,columns=['Name','Age'], index=[1,2,3])
print(type(list_df))
print(list_df)
```

```python
my_Tuple = (('Alice', 25),('Bob', 30), ('Charlie', 28))

tuple_df = pd.DataFrame(my_Tuple, columns=['Name', "Age"], index=[1,2,3])
print(tuple_df)
```

```python
my_dict = {
    "Name": ['Alice','Bob','Charlie'],
    "Age" : [25, 30, 28]
}

dict_df = pd.DataFrame(my_dict, index=[1,2,3])
print(dict_df)
```

```python
# Json Data
My_data = [
    {"Name": "Alice", "Age": 25, "City":"New York"},
    {"Name" : "Bob", "Age":30, "City": "Paris"},
    {"Name": "Charlie", "Age": 28},

]

data_df = pd.DataFrame(My_data, index=[1,2,3])
print(data_df)
```
