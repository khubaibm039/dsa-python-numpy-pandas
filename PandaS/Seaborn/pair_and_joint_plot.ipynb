```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/student_dataset_complete.csv')
df.head()
```

```python
#Pair plot

student_marks = df[['Marks', 'study_hours', 'attendance_rate', 'gender']]
student_marks.head()
```

```python
sns.pairplot(data=student_marks)
```

```python
sns.pairplot(data=student_marks,kind='hist')
```

```python
sns.pairplot(data=student_marks, hue='gender')
```

```python
#joint plot

sns.jointplot(data=student_marks, x='study_hours', y='Marks')
```

```python
sns.jointplot(data=student_marks, x='study_hours', y='Marks', hue='gender')
```

```python
sns.jointplot(data=student_marks, x='study_hours', y='Marks', kind='kde')
```

```python
sns.jointplot(data=student_marks, x='study_hours', y='Marks', kind='kde',hue='gender')
```
