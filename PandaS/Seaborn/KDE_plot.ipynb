```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips_data = sns.load_dataset('tips')
tips_data.head()
```

```python
sns.displot(kind='hist',data=tips_data, x='total_bill',bins=20)
plt.show()
```

```python
sns.displot(kind='hist',data=tips_data, x='tip',bins=10)
plt.show()
```

```python
sns.kdeplot(data=tips_data, x='tip',hue='sex')
```

```python
sns.kdeplot(data=tips_data, x='tip',hue='sex',fill=True,common_grid=True)
plt.show()
```
