```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/sns_data (1).csv')
df.head()
```

```python
sns.lineplot(data= df, x='week', y='attendance_rate',errorbar=None)
plt.show()
```

```python
sns.lineplot(data= df, x='week', y='attendance_rate',errorbar=None,hue='gender')
plt.show()
```

```python
sns.relplot(kind='line',data= df, x='week', y='attendance_rate',errorbar=None,hue='gender',)
plt.show()
```

```python
sns.relplot(kind='line',data= df, x='week', y='test_score',errorbar=None,hue='gender',)
plt.show()
```

```python
sns.relplot(kind='line',data= df, x='week', y='test_score',errorbar=None,hue='class_level',)
plt.show()
```
