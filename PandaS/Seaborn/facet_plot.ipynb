```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/student_dataset_complete.csv')
df.head()
```

```python
sns.scatterplot(data=df, x='study_hours', y='test_score', hue='gender')
plt.show()
```

```python
sns.relplot(kind='scatter',data=df, x='study_hours', y='test_score', hue='gender')
plt.show()
```

```python
sns.relplot(kind='scatter',data=df, x='study_hours', y='test_score', col='gender')
plt.show()
```

```python
sns.relplot(kind='scatter',data=df, x='study_hours', y='test_score', col='gender',row='hostel')
plt.show()
```

```python
sns.relplot(kind='scatter',data=df, x='study_hours', y='test_score', col='week')
plt.show()
```

```python
sns.relplot(kind='scatter',data=df, x='study_hours', y='test_score', col='week', col_wrap=2)
plt.show()
```
