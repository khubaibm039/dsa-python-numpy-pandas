```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/student_dataset_complete.csv')
df.head()
```

```python
sns.regplot(data=df,x='study_hours', y='test_score')
plt.show()
```

```python
sns.lmplot(data=df,x='study_hours', y='test_score')
plt.show()
```

```python
sns.lmplot(data=df,x='study_hours', y='test_score',hue='gender')
plt.show()
```
