```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/student_dataset_complete.csv')
df.head()
```

```python
sns.barplot(data= df, x='gender', y='Marks', errorbar=None)
plt.show()
```

```python
sns.barplot(data= df, x='gender', y='Marks', errorbar=None,estimator=np.median)
plt.show()
```

```python
sns.barplot(data= df, x='gender', y='Marks', errorbar=None,estimator=np.max)
plt.show()
```

```python
sns.barplot(data= df, x='gender', y='Marks', errorbar=None,estimator=np.sum)
plt.show()
```
