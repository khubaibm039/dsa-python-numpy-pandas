```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/sns_data (1).csv')
df.head()
```

```python
sns.scatterplot(data=df, x='study_hours', y='test_score')
plt.show()
```

```python
sns.scatterplot(data=df, x='study_hours', y='test_score', hue='gender')
plt.show()
```

```python
sns.scatterplot(data=df, x='study_hours', y='test_score', hue='gender',style='class_level')
plt.show()
```

```python
sns.scatterplot(data=df, x='study_hours', y='test_score', hue='gender',style='subject')
plt.show()
```

```python
sns.scatterplot(data=df, x='study_hours', y='test_score', hue='gender',size='Tshirt_size')
plt.show()
```

```python
sns.relplot(kind='scatter',data=df, x='study_hours', y='test_score', hue='gender',size='Tshirt_size')
plt.show()
```
