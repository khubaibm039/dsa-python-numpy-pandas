2. Check Duplicates and Remove Them
Column: EmployeeID
 Tasks:
Count duplicate entries


Remove duplicates


Display cleaned DataFrame


```python
#Count duplicate entries
duplicate_entries = len(df['EmployeeID']) - len(df['EmployeeID'].unique())
print('Duplicate Entries: ',duplicate_entries)

#Remove duplicates
remove_duplicate = df.drop_duplicates('EmployeeID')
print(remove_duplicate)

# Display cleaned DataFrame
print(df.dropna())
```

3. Handle Missing Values
Columns: Age, Salary
 Tasks:
Count missing values


Fill missing Age with mean age


Drop rows where Salary is missing


```python
#Count missing values
null_age_values = len(df['Age']) - len(df.dropna(subset='Age'))
null_salary_values = len(df['Salary']) - len(df.dropna(subset='Salary'))

print('Missing age values: ', null_age_values)
print('Missing Salary Values: ', null_salary_values)
print('Total Missing Values:', null_age_values + null_salary_values)

#Fill missing Age with mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
print('\n')
print(df.dropna(subset="Salary"))
```
