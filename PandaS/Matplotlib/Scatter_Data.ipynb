```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./Files/student_IQdata.csv')
df
```

```python
plt.xlabel('Study_Hour')
plt.ylabel('IQ_Score')
plt.title('Student Study Hours and their Iq levels')

plt.scatter(df['Study_Hour'], df['IQ_Score'],color='orange')
plt.grid(linestyle= 'dotted')
```

```python
plt.xlabel('Study_Hour')
plt.ylabel('IQ_Score')
plt.title('Student Study Hours and their Iq levels')

plt.scatter(df['Study_Hour'], df['IQ_Score'],color='orange')
plt.scatter(df['Study_Hour'], df['Chilling_Hours'],color='black')

plt.legend()
plt.grid(linestyle= 'dotted')
```
