```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
hours = [2,3,6,3,5,6,5]
days = [1,2,3,4,5,6,7]

plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Hours')
plt.title('Trend of a Student Study Time Over a Week')

plt.plot(days, hours)
```

```python
hours_1 = [2,3,6,3,5,6,5]
hours_2 = [1,3,5,7,5,2,3]
days = [1,2,3,4,5,6,7]

plt.xlabel('Days')
plt.ylabel('Hours')
plt.title('Trend of Two Student Study Time Over a Week')

plt.plot(days, hours_1)
plt.plot(days, hours_2)
```

```python
df = pd.read_csv('./Files/enrollment_data.csv')
df
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')


plt.plot(df['Year'], df['Programming'])
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'], label= "Programming")
plt.plot(df['Year'], df['Digital Marketing'], label= "Digital Marketing")

plt.legend()
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'], label= "Programming",color='#0008ff')
plt.plot(df['Year'], df['Digital Marketing'], label= "Digital Marketing", color='black')

plt.legend()
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'], label= "Programming",color='#0008ff',linewidth=3,linestyle= 'dotted')
plt.plot(df['Year'], df['Digital Marketing'], label= "Digital Marketing", color='black',linewidth=3,linestyle= 'dashed')

plt.legend()
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'], label= "Programming",color='#0008ff', marker ='.')
plt.plot(df['Year'], df['Digital Marketing'], label= "Digital Marketing", color='black',marker='o')

plt.legend()
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'], label= "Programming",color='#0008ff', marker ='.')
plt.plot(df['Year'], df['Digital Marketing'], label= "Digital Marketing", color='black',marker='o')

plt.legend()
plt.grid()
plt.show()
```

```python
plt.xlabel('Years')
plt.ylabel('Count for Enrollment')
plt.title('Student Enrollment Over 8 The Year')

plt.plot(df['Year'], df['Programming'], label= "Programming",color='#0008ff', marker ='.')
plt.plot(df['Year'], df['Digital Marketing'], label= "Digital Marketing", color='black',marker='o')

plt.legend()
plt.grid(linestyle= 'dotted')
plt.show()
```
