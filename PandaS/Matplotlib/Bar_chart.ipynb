```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
df
```

```python
gender_group = df.groupby('Gender').size()
gender_group
gender_group.index
gender_group.values
```

```python
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('bar chart for male and female')

gender_group = df.groupby('Gender').size()

plt.bar(gender_group.index, gender_group.values, color =['orange', 'black'], edgecolor= 'blue')
```

```python
df = pd.read_csv('../Files/student_data.csv')
df
group_complete_status = df.groupby('CompletionStatus').size()
group_complete_status.index
group_complete_status.values
```
