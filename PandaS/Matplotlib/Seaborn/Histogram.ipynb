```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('./Files/student_dataset_complete.csv')
df.head()
```

```python
sns.histplot(data=df, x='attendance_rate')
plt.show()
```

```python
sns.histplot(data=df, x='attendance_rate',hue='gender')
plt.show()
```

```python
sns.histplot(data=df, x='attendance_rate',hue='gender',bins=10)
plt.show()
```

```python
sns.histplot(data=df, x='attendance_rate',hue='gender',bins=10,element='step')
plt.show()
```

```python
sns.displot(kind='hist',data=df, x='attendance_rate',hue='gender',bins=10,element='step')
plt.show()
```

```python
sns.displot(kind='hist',data=df, x='attendance_rate',col='gender',bins=10,element='step')
plt.show()
```

```python
sns.displot(kind='hist',data=df, x='study_hours',col='gender')
plt.show()
```
