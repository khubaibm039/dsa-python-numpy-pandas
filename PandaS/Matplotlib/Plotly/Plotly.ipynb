```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df = pd.read_csv('./Files/student_dataset_complete.csv')
df.head()
```

```python
fig = px.scatter(df,
                x='time_study',
                y='Marks',
                color='gender',
                size='Tshirt_size',
                hover_data=['time_study', 'Marks', 'hostel', 'gender','attendance_rate','subject']
)
fig.show()
```

```python
enrollment_data = pd.read_csv('./Files/enrollment_data.csv')
enrollment_data
```

```python
fig = px.line(enrollment_data, x='Year', y=['Programming','Digital Marketing'],markers='True')
fig.show()
```
