```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../Files/student_data.csv')
df
group_complete_status = df.groupby('CompletionStatus').size()
group_complete_status
```

```python
plt.pie(group_complete_status,
        labels=group_complete_status.index,
        autopct='%1.1f%%',
        # startangle=90,
        shadow=True,
        explode=(0.06, 0.1, 0)
   )
plt.show()
```
