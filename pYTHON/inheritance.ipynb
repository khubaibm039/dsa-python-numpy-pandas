```python
class Person:
    def __init__(self, fname, lname):
        self.fname= fname
        self.lname= lname
    def printname(self):
        print(self.fname, self.lname)

p1 = Person("John", "Doe")
print(p1.printname())
```

```python
class Student(Person):
    def __init__(self, fname, lname):
        Person.__init__(self,fname, lname)
x = Student("Mike", "Olsen")
print(x.printname())
```

```python
class Student(Person):
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)
        self.graduateYear = year
x = Student("mike", "duo", 2019)
print(x.graduateYear)
```

```python
class Phone :
    def __init__(self, model, battery, camera, battery_percentage = 100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage
    def charge(self, hour):
        print(f"charge complete by {hour}")
    def capture(self, photo):
        if (self.battery_percentage)<=0:
            print("No charge")
        else:
            self.battery_percentage -= photo
            print(f"photo captured in {self.model}")
```

```python
apple = Phone("iphone-17", 3000, 40)
blueberry= Phone("B-19", 4000,30)
motorola = Phone("M-12", 3500, 35)
apple.capture(10)
```
