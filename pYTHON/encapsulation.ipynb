```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age
    def get_age(self):
        return self.__age
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive")

    
p1 = Person("Tobias", 25)
p1.set_age(2165)
print(p1.get_age())
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.__grade = 0

    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Grade must be between 0 to 100")
    def get_grade(self):
        return self.__grade

    def status_grade(self):
        if self.__grade >= 60:
            return "Passed"
        else:
            return "Failed"
s1 = Student("Khali")
s1.set_grade(98)
print(s1.name,s1.get_grade(),s1.status_grade())
print()
```
