```python
import numpy as np

x = np.random.randint(1,10,size=5)
print(x)
sin = np.sin(x)
print(sin)

radian_to_deg = np.rad2deg(x)
print(radian_to_deg)
deg_2_rad = np.deg2rad(x)
print(deg_2_rad)
```

```python
base_10_log = np.log10(x)
print(base_10_log)
```

```python
sqrt = np.sqrt(x)
print(sqrt)
```
