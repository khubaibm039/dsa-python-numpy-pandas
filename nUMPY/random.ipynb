```python
import numpy as np

arr = np.random.rand(2, 3, 1) # 0 -> 1

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.base)
print(arr.size)
print(arr.ndim)
```

```python
import numpy as np
# np.random.randint(start, end , shape)
arr = np.random.randint(-10,10 ,(10, 10)) # -10 -> 10

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.base)
print(arr.size)
print(arr.ndim)
```

```python
import numpy as np
# np.random.uniform(start, end , shape)
arr = np.random.uniform(0,2 ,(3,3, 4)) # -10 -> 10

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.base)
print(arr.size)
print(arr.ndim)
```

```python
arr = np.arange(1, 13, 1).reshape(2,2,3)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.base)
print(arr.size)
print(arr.ndim)
```

```python
arr = np.linspace(1, 10, 15)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.base)
print(arr.size)
print(arr.ndim)
```

```python
arr = np.logspace(0, 4, 6, base=2)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.base)
print(arr.size)
print(arr.ndim)
```
