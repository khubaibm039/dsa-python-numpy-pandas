```python
import numpy as np
arr = np.array([10,8,16,100])
x2 = np.array([[10,3,5],[3,4,9]])
print(arr)
print(x2)
```

```python
index = np.where(arr == 8)
print(index)
```

```python
index = np.where(arr > 8, 0, arr)
print(index)
```

```python
index = np.where(x2 > 8)
print(index)
```

```python
index = np.where(x2 > 8, x2, 0)
print(index)
```

```python
print(arr)
maximum_index = np.argmax(arr)
print(maximum_index)
```

# Counting

```python
c = np.random.randint(1,100, size=(100,))
print(c)
```

```python
value_grater_than_60 = np.count_nonzero(c < 70)
print(value_grater_than_60)
```

```python
unique_value = np.unique(c, return_counts=True)
print(unique_value[1])
```
