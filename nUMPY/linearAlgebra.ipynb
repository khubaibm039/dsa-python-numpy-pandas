```python
import numpy as np
study_hours = np.array([2,4,5,7,8])
exam_scores = np.array([65,75,78,88,92])

data = np.array([study_hours, exam_scores])

corelation = np.corrcoef(data)

print(corelation)
```

# linear Algebra

```python
A = np.array([
    [1,2,3],
    [4,5,6]
])
B = np.array([
    [7,  8 ],
    [9,  10],
    [11, 12]
])

dot_product = np.dot(A,B)
print(dot_product) 
print(np.trace(B))
```

```python
import numpy as np

sq_mat = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 11]
])

det_of_sq = np.linalg.det(sq_mat)
print(det_of_sq)
rank_sq = np.linalg.matrix_rank(sq_mat)
print(rank_sq)
```
